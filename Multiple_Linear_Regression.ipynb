```python
import numpy as np
```

```python
class MyLR:
    def _init_(self):
        self.coef = None
        self.intercept = None
    def fit(self, X_train, y_train):
        X_train = np.insert(
            X_train, 0, 1, axis=1
        )
        beta = np.linalg.inv(
            X_train.T @ X_train
        ) @ X_train.T @ y_train
        self.intercept = beta[0]
        self.coef = beta[1:]
    def predict(self, X_test):
        return X_test @ self.coef + self.intercept
```

```python
my_lr = MyLR()
```

```python
my_lr.fit(X_train, y_train)
```

```python
my_y_pred = my_lr.predict(X_test)
```

```python
print ('My lr R2 Score:', r2_score(y_test, my_y_pred))
print('My lr Coefficient:', my_lr.coef)
print('My lr Intercept:', my_lr.intercept)
```
