```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df.drop(columns='churned'), df['churned'],
    test_size=0.2, random_state=42)
```

```python
X_train = X_train.drop(
    columns=['customer_id', 'customer_name', 'signup_date']
)

X_test = X_test.drop(
    columns=['customer_id', 'customer_name', 'signup_date']
)
```

```python
X_train = pd.get_dummies(X_train, columns=['region'], dtype=int)
X_test = pd.get_dummies(X_test, columns=['region'], dtype=int)

X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
scaler =  StandardScaler()
```

```python
X_train.dtypes
```

```python
X_train.select_dtypes(exclude='number').columns
```

```python
X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)
```

```python
import numpy as np
```

```python
print("NaN values:",
np.isnan(X_train_scaled).sum())
print("Infinite values:",
     np.isinf(X_train_scaled).sum())
print("Shape",
     X_train_scaled.shape)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()
```

```python
X_trained_PCA = pca.fit(X_train_scaled)
```

```python
pca.n_components_
```

```python
pca.explained_variance_ratio_
```

```python
np.cumsum(pca.explained_variance_ratio_)
```

```python
pca_ = PCA(n_components=0.95)
```

```python
X_trained_PCA_ = pca_.fit_transform(X_train_scaled)
X_tested_PCA_ = pca_.transform(X_test_scaled)
```

```python
print(X_trained_PCA_.shape)
print(X_tested_PCA_.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
knn_original = KNeighborsClassifier(n_neighbors=5)
knn_original.fit(X_train_scaled, y_train)

y_pred_original = knn_original.predict(X_test_scaled)
```

```python
knn_pca = KNeighborsClassifier(n_neighbors=5)
knn_pca.fit(X_trained_PCA_, y_train)

y_pred_pca = knn_pca.predict(X_tested_PCA_)
```

```python
print("Without PCA:", accuracy_score(y_test, y_pred_original))
print("With PCA:", accuracy_score(y_test, y_pred_pca))
```

```python
from sklearn.metrics import classification_report

print("Without PCA:")
print(classification_report(y_test, y_pred_original, zero_division=0))

print("With PCA:")
print(classification_report(y_test, y_pred_pca, zero_division=0))
```
